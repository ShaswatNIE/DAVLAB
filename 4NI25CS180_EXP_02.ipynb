```python
import pandas as pd
import numpy  as  np
```

```python
data={
    'Name':['Ankit','Bebo','Chetan','Amit','Divya','Ravi'],
    'Age':[25,np.nan,56,45,12,78],
    'Marks':[85,45,63,np.nan,69,67],
    'City':['Mumbai','Delhi','Bangalore','Chennai','Pune',np.nan]
}
df=pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\n Shape of dataset(rows,coulmns):",df.shape)
print("\n column names:",df.columns.tolist)
print("\n data types of each column")
print(df.dtypes)
print("\dataset info:")
df.info()
```

```python
print("\nDescriptive Statistic(numerical columns):")
print(df.describe())
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:\n",df.median(numeric_only=True))
print("\nMode:\n",df.mode(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\misssing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in the dataset:",df.isnull().sum().sum)
```

```python
print("\n duplicate rows in dataset:")
print(df[df.duplicated()])
print("\nTotal duplicate rows in the dataset:",df.duplicated().sum())
print("\nEDA process is completed")
```
