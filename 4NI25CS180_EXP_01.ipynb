```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print("Numpy Versions",np.__version__)
print("Pandas Versions",pd.__version__)
```

```python
marks=np.array([78,56,45,65,29])
print("Marks:",marks)
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
print("Type:",type(marks))
```

```python
print("Numbers of dimensions:",marks.ndim)
print("Shape:",marks.shape)
print("Size:",marks.size)
```

```python
print("Original:",marks)
print("ADD 5:",marks+5)
print("Multiply by 2:",marks*2)
print("divide by 2:",marks/2)
print("Square:",marks**2)
print("Sum:",marks.sum())
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
```

```python
print("First mark:",marks[0])
print("Second mark:",marks[1])
print("Last mark:",marks[-1])
```

```python
print("Index 1 to 3:",marks[1:4])
print("First three:",marks[:3])
print("From index two:",marks[2:])
print("Eevery second element",marks[::2])
```

```python
student_marks=np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(student_marks)
print("shape:",student_marks.shape)
```

```python
print("Mean of all Value:",student_marks.mean())
print("Mean of each column:",student_marks.mean(axis=0))
print("Mean of each row:",student_marks.mean(axis=1))
```

```python
import numpy as np
import pandas as pd

data1 = {
    "Name": ["shaswat", "shivansh", "ssk", "shweta", "shreyas"],
    "Branch": ["cse", "ise", "cse", "aiml", "ise"],
    "Marks": [72, 91, 84, 95, 68],
    "Attendance": [82, 94, 88, 96, 75]
}

df1 = pd.DataFrame(data1)

print(df1[["Name", "Marks"]])
print(df1[df1["Marks"] > 70])
print(df1[df1["Attendance"] > 85])
print("First three:", df1[:3])

print("Mean:", df1["Marks"].mean())
print("Maximum:", df1["Marks"].max())
print(df1["Branch"]=="cse")
display(
    df1[
        (df1["Branch"] == "cse") &
        (df1["Attendance"] > 85)
    ]
)









```
